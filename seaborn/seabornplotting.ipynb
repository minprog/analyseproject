Deze dataset is overgenomen van de grafieken uit dr. Stat. Als voorbeeld bekijken wij hoe je deze grafieken zelf kan maken. 

Hiervoor gebruiken wij de [_sns.histplot()_](https://seaborn.pydata.org/generated/seaborn.histplot.html#seaborn.histplot) functie. Als je even spiekt bij de documentatie kan het intimiderend zijn hoeveel variabelen er met deze functie mee kunnen. Maar, in de realiteit is het juist deze opties in een simpele functie die het maken van een plot makkelijker maken. 

Om te beginnen willen we de volgend [_afbeelding_](https://drstat.net/nl/pages/02-tabellen_en_grafieken/53) namaken. 

Als we simpeleg de data invoeren krijgen we nog niet helemaal wat we willen....

```python
plot0 = sns.histplot( data=woningen, x="woonoppervlak (m2)")
```

Zoals je ziet komt het al enigszins in de buurt, maar is de data anders verdeeld. Bij het maken van deze histogram hebben we niet netjes aangegeven wat het klassenmidden, en de klassengrenzen zijn van elk van onze staven. Dit kunnen we meegeven aan de functie aan de hand van de _bins_ (het aantal klassen) en de _binwidth_ (de breedte van elke klasse). 

```python
plot1 = sns.histplot( data=woningen, x="woonoppervlak (m2)", binwidth=5, bins=8)
```

Nu zijn we er bijna. Alleen, begint onze eerste bin nu bij het laagste getal in onze dataset (3). Wij hebben liever dat deze begint bij nul. Dit geef je mee met de _binrange()_. __Let op:__ bij het gebruik van binrange worden alle datapunten buiten de range genegeerd. Denk bij je eigen data na of dit de bedoeling is of niet. 

```python
plot2 = sns.histplot( data=woningen, x="woonoppervlak (m2)", binwidth=5, bins=8, binrange=(0, 40))
```

### Histogram weergeven als een frequentiepolygoon

Nu hebben we een nette histogram! Maar we willen uiteraard ook de andere twee vormen van deze data weergeven; het [_frequentiepolygoon_](https://drstat.net/nl/pages/02-tabellen_en_grafieken/56) en het [_cumulatief frequentiepolygoon_](https://drstat.net/nl/pages/02-tabellen_en_grafieken/57). 

Op de [Seaborn website](https://seaborn.pydata.org/generated/seaborn.histplot.html#seaborn.histplot) kunnen we deze variabelen vinden: 

__element : {“bars”, “step”, “poly”}__ <br>
Visual representation of the histogram statistic. Only relevant with univariate data.

__fill : bool__ <br>
If True, fill in the space under the histogram. Only relevant with univariate data.

Als we deze toevoegen aan onze functie, krijgen we een frequentiepolygoon zoals wij deze verwachten. 

```python
plot2 = sns.histplot( data=woningen, x="woonoppervlak (m2)", binwidth=5, bins=8, binrange=(0, 40), element="poly", fill=False)
```

Voor de derde grafiek kunnen we nog een enkele optie toevoegen. 

__cumulative : bool__ <br>
If True, plot the cumulative counts as bins increase.

```python
plot3 = sns.histplot( data=woningen, x="woonoppervlak (m2)", binwidth=5, bins=8, stat="count", binrange=(0, 40), element="poly", fill=False, cumulative=True)
```

### Data opdelen in kolommen, rijen en kleuren

Voor de volgende voorbeelden gebruiken we een test dataset die gaat over de fooi die wordt gegeven bij het uit eten gaan.

```python
tips = sns.load_dataset("tips")
sns.histplot( data=tips, x="tip", element="poly")

display(tips)
```

Hierboven hebben we telkens één kolom aan data gebruikt. Maar, in een ander gevallen zal je misschien twee kolommen data tegen elkaar uit willen spelen. Seaborn kan je data op verschillende manieren splitsen. Om dit te doen gebruiken we net een andere functie: <br>
ipv __sns.histplot()__ gebruiken we __sns.displot( kind="hist")__.

De [displot functie](https://seaborn.pydata.org/generated/seaborn.displot.html#seaborn.displot) laat ons de data splitsen op een aantal vlakken: <br>
__hue : vector or key in data__ <br>
Semantic variable that is mapped to determine the color of plot elements.

__row, col : vectors or keys in data__ <br>
Variables that define subsets to plot on different facets.

```python
sns.displot(data=tips, kind="hist", x="tip", element="poly", hue="sex", col="time", row="smoker")
```

## Scatterplots

Als laatste geven we een voorbeeld van een plot die jullie morgen bij het hoofdstuk Correlatie van dr. Stat veel zullen zien: <br>
een [Scatter plot](https://seaborn.pydata.org/generated/seaborn.scatterplot.html#seaborn.scatterplot). 

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

Ook hier kunnen we zo nodig met een angepaste functie een groep aan grafieken maken: <br>
ipv __scatterplot()__ gebruiken we __relplot( kind="scatter")__

```python
sns.relplot( kind="scatter", data=tips, x="total_bill", y="tip", hue="smoker", col="time") 
```

Je zou kunnen opmerken dat als we de lunch en het diner uitelkaar zetten, dat er een aantal nieuwe inzichten tevoren komen. Zo zien wij bijvoorbeeld:

1. Er zijn meer datapunten bij diner dan bij lunch
2. De total bill is bij diner gemiddeld hoger, maar bevat daar ook het merendeel aan de uitschieters, zowel over de correlatie lijn (links onder en recht boven) als de uitschieters die ver van de correlatie lijn af liggen (rechts onder en links boven.
3. Er zijn in de data een aantal waardes die significant meer voorkomen in de tips (ronde getallen). We kunnen door de dichtheid van de punten in de scatterplot zien dat deze bij 2 en 3 liggen, maar bij de lunch enkel bij de 2. Als we dit vervolgens in een histogram uitzetten zien we dat nog duidelijker.

```python
sns.displot( kind="hist", data=tips, x="tip", col="time", binwidth=0.1) 
```
